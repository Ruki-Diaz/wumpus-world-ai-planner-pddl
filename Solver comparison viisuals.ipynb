# Solver Comparison for Wumpus World (SIT215 Assignment 3)
This notebook visualizes and compares the performance of two PDDL solvers,
- **BFWS (FF Parser)**
- **LAMA-First (Fast Downward)**

The metrics include planning time, nodes generated/expanded, plan length, and peak memory usage.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Solver performance data
data = {
    "Solver": ["BFWS (FF Parser)", "LAMA-First"],
    "Planning Time (s)": [0.000456, 0.027971],
    "Nodes Generated": [89, 52],
    "Nodes Expanded": [58, 12],
    "Plan Length": [10, 10],
    "Peak Memory (KB)": [None, 12500]
}

df = pd.DataFrame(data)
sns.set(style="whitegrid")
```

```python
# Plot 1: Planning Time
plt.figure(figsize=(8, 4))
sns.barplot(x="Solver", y="Planning Time (s)", data=df)
plt.title("Planning Time Comparison")
plt.ylabel("Time (seconds)")
plt.xlabel("Solver")
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Nodes Generated vs Expanded
plt.figure(figsize=(8, 4))
nodes_df = df.melt(id_vars="Solver", value_vars=["Nodes Generated", "Nodes Expanded"],
                   var_name="Metric", value_name="Count")
sns.barplot(x="Solver", y="Count", hue="Metric", data=nodes_df)
plt.title("Nodes Generated vs Expanded")
plt.ylabel("Number of Nodes")
plt.xlabel("Solver")
plt.tight_layout()
plt.show()
```

```python
# Plot 3: Plan Length
plt.figure(figsize=(8, 4))
sns.barplot(x="Solver", y="Plan Length", data=df)
plt.title("Plan Length Comparison")
plt.ylabel("Steps")
plt.xlabel("Solver")
plt.tight_layout()
plt.show()
```
